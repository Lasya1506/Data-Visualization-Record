```python
import time
import numpy as np

# Define a large size for comparison
size = 10**6

# --- Python List Performance ---
start_time = time.time()
python_list = list(range(size))
python_list_creation_time = time.time() - start_time

start_time = time.time()
python_list_sum = [x + 1 for x in python_list] # Example operation
python_list_operation_time = time.time() - start_time

print(f"Python List Creation Time: {python_list_creation_time:.6f} seconds")
print(f"Python List Operation Time (element-wise +1): {python_list_operation_time:.6f} seconds")

print("\n" + "-"*30 + "\n")

# --- NumPy Array Performance ---
start_time = time.time()
numpy_array = np.arange(size)
numpy_array_creation_time = time.time() - start_time

start_time = time.time()
numpy_array_sum = numpy_array + 1 # Example operation
numpy_array_operation_time = time.time() - start_time

print(f"NumPy Array Creation Time: {numpy_array_creation_time:.6f} seconds")
print(f"NumPy Array Operation Time (element-wise +1): {numpy_array_operation_time:.6f} seconds")
```

```python
import sys
import numpy as np

# Assuming 'size' is already defined from the previous cell execution (e.g., 10**6)
# If not, uncomment the line below:
# size = 10**6

# --- Python List Memory Usage ---
python_list = list(range(size)) # List of integers
# sys.getsizeof() returns the size of the list object itself, not its contents.
# To get the approximate total memory, we also need to account for each element.
# On a 64-bit system, an integer object typically takes 28 bytes.
python_list_memory_bytes = sys.getsizeof(python_list) + size * sys.getsizeof(python_list[0])
python_list_memory_mb = python_list_memory_bytes / (1024 * 1024)

print(f"Python List (size={size}) memory: {python_list_memory_mb:.2f} MB")

print("\n" + "-"*30 + "\n")

# --- NumPy Array Memory Usage ---
numpy_array = np.arange(size) # Array of integers (default dtype is usually int64)
numpy_array_memory_bytes = numpy_array.nbytes
numpy_array_memory_mb = numpy_array_memory_bytes / (1024 * 1024)

print(f"NumPy Array (size={size}, dtype={numpy_array.dtype}) memory: {numpy_array_memory_mb:.2f} MB")

# Demonstrate with a smaller dtype for NumPy to show further optimization
numpy_array_int32 = np.arange(size, dtype=np.int32)
numpy_array_int32_memory_bytes = numpy_array_int32.nbytes
numpy_array_int32_memory_mb = numpy_array_int32_memory_bytes / (1024 * 1024)

print(f"NumPy Array (size={size}, dtype={numpy_array_int32.dtype}) memory: {numpy_array_int32_memory_mb:.2f} MB")
```
