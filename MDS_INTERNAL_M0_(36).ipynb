```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/sample_data/california_housing_train.csv')

plt.figure(figsize=(10, 5))
plt.plot(df['median_house_value'].head(100), color='blue')
plt.title('Line Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/sample_data/california_housing_train.csv')

df['age_group'] = pd.cut(df['housing_median_age'], bins=5)


age_means = df.groupby('age_group')['median_house_value'].mean()

plt.figure(figsize=(8, 5))
age_means.plot(kind='bar', color='orange')
plt.title('barplot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/sample_data/california_housing_train.csv')

plt.figure(figsize=(8, 5))
plt.hist(df['median_income'], bins=30, color='green',edgecolor="black")
plt.title('histogram')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/sample_data/california_housing_train.csv')


room_sample = df['total_rooms'].sample(1000)

plt.figure(figsize=(8, 5))
room_sample.plot(kind='density', color='red')
plt.title('Density Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.xlim(-5000, 20000)
plt.show()
```
